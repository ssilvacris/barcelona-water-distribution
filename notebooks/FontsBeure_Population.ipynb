## 2. Load Datasets

```python
# Barcelona Population
pop = pd.read_csv("poplacio_2019_2024.csv")

# Barcelona Fountains
fountains = pd.read_csv("fonts_beure_2019_2024.csv")

# Barcelona Districts
district_ref = pd.read_csv("BarcelonaCiutat_Districtes.csv")
```

```python
display(pop.head())
display(fountains.head())
display(district_ref.head())
```

```python
#Rename key columns

pop = pop.rename(columns={
    "Codi_Districte": "Codi_Districte",
    "Nom_Districte": "DISTRICT_NAME",
    "Valor": "population"
})
district_ref = district_ref.rename(columns={
    "Codi_Districte": "Codi_Districte",
    "Nom_Districte": "DISTRICT_NAME"
})
```

```python
# CODI looks like "01-004" → we only need "01"
fountains["CODI"] = fountains["CODI"].astype(str)
fountains["Codi_Districte"] = fountains["CODI"].str.split("-").str[0]

# your pop/district_ref use codes without leading zero → convert all to int→str
pop["Codi_Districte"] = pop["Codi_Districte"].astype(int).astype(str)
district_ref["Codi_Districte"] = district_ref["Codi_Districte"].astype(int).astype(str)
fountains["Codi_Districte"] = fountains["Codi_Districte"].astype(int).astype(str)
```

```python
# population per year + district
pop_year = (
    pop.groupby(["year", "Codi_Districte", "DISTRICT_NAME"], as_index=False)
       .agg({"population": "sum"})
)

# fountains per year + district code
fount_year = (
    fountains.groupby(["year", "Codi_Districte"], as_index=False)
             .size()
             .rename(columns={"size": "fountains"})
)
```

```python
# join only on columns that exist in both
df = pop_year.merge(
    fount_year[["year", "Codi_Districte", "fountains"]],
    on=["year", "Codi_Districte"],
    how="left"
)

# now DISTRICT_NAME comes from pop_year
df["fountains_per_1000"] = (df["fountains"] / df["population"] * 1000).round(3)

df = df.sort_values(["DISTRICT_NAME", "year"]).reset_index(drop=True)
df.head(12)
```

```python
# salve the combined file

df.to_csv("df_2019_2024.csv", index=False)
```

## 4. Interactive Map with Folium

### 4.1 All fountains as points

```python
# Center of Barcelona
barcelona_center = [41.3874, 2.1686]

m = folium.Map(location=barcelona_center, zoom_start=12, tiles="CartoDB positron")

marker_cluster = MarkerCluster().add_to(m)

for _, row in fountains.iterrows():
    lat = row.get("LATITUD") or row.get("lat") or row.get("Latitude")
    lon = row.get("LONGITUD") or row.get("lon") or row.get("Longitude")
    # skip if missing
    if pd.isna(lat) or pd.isna(lon):
        continue
    
    popup_text = f"District: {row.get('DISTRICT_NAME', 'N/A')}<br>Year: {row.get('year', 'N/A')}"
    folium.Marker(
        location=[lat, lon],
        popup=popup_text,
        icon=folium.Icon(color="blue", icon="tint", prefix="fa")
    ).add_to(marker_cluster)

m
```

### 4.2 Filter the map by year

```python
# center on Barcelona
barcelona_center = [41.3874, 2.1686]

# base map
m = folium.Map(location=barcelona_center, zoom_start=12, tiles="OpenStreetMap")

# list of years in your fountains file
years = sorted(fountains["year"].dropna().unique())

for yr in years:
    # create a layer for that year
    fg = folium.FeatureGroup(name=str(yr), show=(yr == 2024))  # show only 2024 by default

    # marker cluster INSIDE the layer
    mc = MarkerCluster(name=f"Fountains {yr}").add_to(fg)

    # filter fountains for that year
    data_year = fountains[fountains["year"] == yr]

    for _, row in data_year.iterrows():
        lat = row.get("LATITUD")
        lon = row.get("LONGITUD")
        if pd.isna(lat) or pd.isna(lon):
            continue

        popup_text = (
            f"District code: {row.get('Codi_Districte', 'N/A')}<br>"
            f"Year: {yr}"
        )

        folium.Marker(
            location=[lat, lon],
            popup=popup_text,
        ).add_to(mc)

    # add the whole layer (with the cluster) to the map
    fg.add_to(m)

# more base layers
folium.TileLayer("CartoDB positron").add_to(m)
folium.TileLayer("Stamen Terrain").add_to(m)

# this creates the checkbox/radio panel
folium.LayerControl(collapsed=False).add_to(m)
m
```

## 7. Simple ML: can population explain fountains?

```python
#Build dataset
ml_data = df.dropna(subset=["population", "fountains"]).copy()

X = ml_data[["population"]]
y = ml_data["fountains"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Train linear regression
model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print(f"R²: {r2_score(y_test, y_pred):.3f}")
print(f"MAE: {mean_absolute_error(y_test, y_pred):.2f}")
print("Intercept:", model.intercept_)
print("Coefficient (population):", model.coef_[0])

# plot actual vs predicted
plt.figure(figsize=(8,6))
plt.scatter(X_test, y_test, label="Actual", alpha=0.7)
plt.scatter(X_test, y_pred, label="Predicted", alpha=0.7, marker="x")
plt.title("Actual vs. Predicted fountains")
plt.xlabel("Population")
plt.ylabel("Fountains")
plt.legend()
plt.show()
```

## 8. Spatial Clistering of Fountains using K-Means

```python
# Ensure lat/lon columns exist and clean
coords = fountains[["LATITUD", "LONGITUD"]].to_numpy()

print("Number of fountains with coordinates:", coords.shape[0])
```

```python
# drop rows without coordinates, but keep them in the same dataframe
fountains = fountains.dropna(subset=["LATITUD", "LONGITUD"]).copy()

coords = fountains[["LATITUD", "LONGITUD"]].to_numpy()

# choose number of clusters
k = 6
kmeans = KMeans(n_clusters=k, random_state=42)
fountains["cluster"] = kmeans.fit_predict(coords)

# cluster centers (we'll draw them on the map)
centers = kmeans.cluster_centers_
```

```python
# Range of possible k values
k_values = range(1, 11)
inertia = []

# Run KMeans for each k and store inertia
for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

# Plot the Elbow Curve
plt.figure(figsize=(8,5))
plt.plot(k_values, inertia, marker='o')
plt.title('Elbow Method to Determine Optimal k')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia (Sum of Squared Distances)')
plt.grid(True)
plt.show()
```

```python
# choose number of clusters
k = 3
kmeans = KMeans(n_clusters=k, random_state=42)
fountains["cluster"] = kmeans.fit_predict(coords)

# cluster centers (we'll draw them on the map)
centers = kmeans.cluster_centers_
```

```python
# Visulaize the clusters on Folium Map

# colors for clusters
cluster_colors = [
    "red", "blue", "green", "purple", "orange", "darkblue", "darkred", "cadetblue"
]

# base map
m = folium.Map(location=[41.3874, 2.1686], zoom_start=12)

# base layers (these show as radio buttons)
folium.TileLayer("openstreetmap").add_to(m)
folium.TileLayer("CartoDB positron", name="cartodbpositron").add_to(m)
folium.TileLayer("Stamen Terrain", name="stamenterrain").add_to(m)

# years available in fountains
years = sorted(fountains["year"].dropna().unique())

for yr in years:
    # one overlay per year (checkbox)
    fg = folium.FeatureGroup(name=str(yr), show=(yr == 2024))
    data_year = fountains[fountains["year"] == yr]

    for _, row in data_year.iterrows():
        lat = row["LATITUD"]
        lon = row["LONGITUD"]
        if pd.isna(lat) or pd.isna(lon):
            continue

        cluster_id = int(row["cluster"])
        color = cluster_colors[cluster_id % len(cluster_colors)]

        popup_text = (
            f"<b>Year:</b> {yr}<br>"
            f"<b>Cluster:</b> {cluster_id}"
        )
        # if you already merged district name earlier, you can add:
        # f"<b>District:</b> {row.get('DISTRICT_NAME','N/A')}<br>"

        folium.CircleMarker(
            location=[lat, lon],
            radius=3,
            color=color,
            fill=True,
            fill_opacity=0.7,
            popup=popup_text
        ).add_to(fg)

    fg.add_to(m)

# add cluster centers (same for all years)
for i, c in enumerate(centers):
    folium.Marker(
        location=[c[0], c[1]],
        icon=folium.Icon(color="black", icon="star"),
        popup=f"Cluster center {i}"
    ).add_to(m)

# panel like your screenshot
folium.LayerControl(collapsed=False).add_to(m)

m

```
